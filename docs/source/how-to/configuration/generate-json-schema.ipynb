# Generate JSON Schemas

This guide shows how to use the `SchemaRegistry` to generate JSON Schemas.

## Create the Registry

First create the registry and register the classes you want included in the schema.

Here the discovery function is used to register all classes in `pyaml` and other packages which define entry points.

```python
from pyaml.validation import SchemaRegistry

registry = SchemaRegistry()
registry.discover()
print(registry)
```

## Generate JSON Schema

Use the `SchemaGenerator` to generate a JSON Schema to use with external tools.

If a base schema has registered concrete or virtual subclasses, the generated JSON Schema includes those alternatives. This allows editors and other JSON Schema tools to offer the appropriate fields for each configuration type.

```python
from pprint import pprint

from pyaml.validation import SchemaGenerator

json_schema = SchemaGenerator.generate("pyaml.magnet.quadrupole.Quadrupole")
pprint(json_schema)
```

The result can also be saved directly to a file.

```python
SchemaGenerator.save("pyaml.magnet.quadrupole.Quadrupole","quadrupole-schema.json")
```
